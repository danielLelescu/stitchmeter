```python
forces = [[2,6,12],
          [4,16,48],
          [16,48,96]]
xs = [4,2,2]

for i in range(0,3):
    ks = []
    ks.append(forces[i][0]/xs[0])
    ks.append((forces[i][1]-(ks[0]*(xs[0]+xs[1])))/xs[1])
    ks.append((forces[i][2]-(ks[0]*(xs[0]+xs[1]+xs[2]))-(ks[1]*(xs[1]+xs[2])))/xs[2])
    #print(ks)
    print([(kN*1000) for kN in ks])
```
